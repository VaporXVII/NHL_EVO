```sql
create or replace view nhl_evo.games.shift_data
comment 'NHL EVO gold layer for game level shift data' as (


    select 
        season, 
        game_id,
        game_date,
        team_id, 
        team_abbrev,
        player_id, 
        player_name,
        shift_id,
        period, 
        start_time, 
        end_time, 
        coalesce(shift_duration, cast('00:00' as string)) as shift_duration,
        shift_number,
        detail_code,
        event_description,
        event_details,
        event_number,
        type_code,
        scrape_ts_utc,
        insert_dte,
        py_source 
    from nhl_evo_staged.games.shift_data 
)
;
```
